## 1. Variables and Data Types

```python
age = 20
gpa = 3.5
name = "Umer Hayatt"
is_student = True
nothing = None
print(type(age), type(gpa), type(name), type(is_student), type(nothing))
```

2- Type conversion with input()

```python
num = int(input("Enter a number: "))
price = float(input("Enter a price: "))
print("Double:", num * 2)
print("Price with tax:", price * 1.1)
```

3: Operators

```python
a, b = 17, 5
print(a + b, a - b, a * b, a / b)
print(a // b, a % b, a ** b)
print(a > b, a == b, a != b)
print(a > 10 and b < 10, a > 20 or b < 10, not (a > b))
print(3 in [1, 2, 3], "x" not in "hello")
x = [1, 2]; y = [1, 2]; z = x
print(x == y, x is y, x is z)
```

4: if / elif / else

```python
marks = int(input("Enter marks: "))
if marks >= 90:
    print("A")
elif marks >= 80:
    print("B")
elif marks >= 70:
    print("C")
else:
    print("Needs improvement")
```

5: Loops, break, continue

```python
for i in range(1, 6):
    print(i)

for i in range(0, 20, 5):
    print(i)

count = 0
while count < 5:
    print("count:", count)
    count += 1

for i in range(1, 11):
    if i == 3:
        continue
    if i == 8:
        break
    print(i)
```

6: Strings

```python
s = "  hello python world  "
print(s.strip())
print(s.strip().title())
print(s.strip().split())
print(s.replace("python", "AI"))

t = "Bootcamp"
print(t[0], t[-1], t[0:4], t[::-1])
```

7: Lists, tuples, sets, dictionaries

```python
fruits = ["apple", "banana"]
fruits.append("mango")
fruits.remove("banana")
print(fruits, len(fruits))

point = (3, 4)
print(point[0])

nums = {1, 2, 2, 3, 3}
nums.add(10)
print(nums)

student = {"name": "Umer Hayatt", "marks": 85}
student["age"] = 19
for key, value in student.items():
    print(key, ":", value)
```

8: Functions

```python
def square(n):
    return n * n

def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

def is_even(n):
    return n % 2 == 0

print(square(6), greet("Umer Hayatt"), is_even(10))
```

Extend the Student Marks Analyzer

```python
marks = [78, 92, 65, 88, 45, 90]

total = sum(marks)
average = total / len(marks)

print("Total:", total)
print("Average:", round(average, 2))
print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))
print("Passed (>=50):", len([m for m in marks if m >= 50]))
```
