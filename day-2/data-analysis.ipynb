## Custom Titanic question

How many female passengers in third class who embarked at Southampton survived?

Filter `Sex`, `Pclass`, `Embarked` and `Survived`, then count the matching rows. This adapted question is carried over from the source practice notebook.

```python
q1 = df.loc[(df["Sex"] == "female") & (df["Pclass"] == 3) &
            (df["Embarked"] == "S") & (df["Survived"] == 1)]
print("Number of survivors matching criteria:", len(q1))
```

### Finding

**33 passengers** match all four conditions in the included dataset. `Embarked == "S"` means Southampton; rows without that embarkation value are excluded.
